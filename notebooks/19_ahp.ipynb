# 19. AHP chọn mô hình triển khai, CPU (GĐ I, task 7.1, 7.2)
Clone `main`, chạy `pytest -q`, rồi chạy `src/ahp.py` trên `reports/tables/summary_test.csv` (số test đã có của GĐ H, không chạy lại test) và
`docs/ahp_matrix.json` (ma trận Phúc đã duyệt 16/11/2026). Ghi `ahp_result.json` và 3 hình `gd7_ahp_*.png`.

```python
import json, os, resource, subprocess, sys, time
from contextlib import contextmanager

WORK = "/kaggle/working"
REPO = "/tmp/gnn-iot-ids"
URL = "https://github.com/minucedev/gnn-iot-ids.git"
INFO = {"steps_s": {}, "device": "CPU", "gpu": None}
T_START = time.time()


def cpu_model():
    try:
        for line in open("/proc/cpuinfo"):
            if line.startswith("model name"):
                return line.split(":", 1)[1].strip()
    except OSError:
        pass
    return "không rõ"


def peak_ram_gb():
    kb = max(resource.getrusage(resource.RUSAGE_SELF).ru_maxrss, resource.getrusage(resource.RUSAGE_CHILDREN).ru_maxrss)
    return round(kb / 2**20, 2)


def save_info():
    INFO["total_s"] = round(time.time() - T_START, 1)
    INFO["peak_ram_gb"] = peak_ram_gb()
    INFO["gpu_hours"] = 0
    with open(f"{WORK}/run_info.json", "w", encoding="utf-8") as f:
        json.dump(INFO, f, ensure_ascii=False, indent=1)


@contextmanager
def step(name):
    t0 = time.time()
    try:
        yield
    finally:
        INFO["steps_s"][name] = round(time.time() - t0, 1)
        save_info()
        print(f"--- {name}: {INFO['steps_s'][name]} s, RAM đỉnh tới nay {INFO['peak_ram_gb']} GB")


def run(cmd, ok=(0,), **kw):
    r = subprocess.run(cmd, capture_output=True, text=True, **kw)
    print(r.stdout[-9000:], end="")
    print(r.stderr[-2500:], end="", file=sys.stderr)
    print(f"[mã thoát {r.returncode}]")
    if r.returncode not in ok:
        raise RuntimeError(f"Lệnh lỗi: {' '.join(cmd)}")
    return r


with step("clone_va_cai_goi"):
    run(["rm", "-rf", REPO])
    run(["git", "clone", "--depth", "1", URL, REPO])
    INFO["commit"] = subprocess.run(["git", "-C", REPO, "rev-parse", "HEAD"], capture_output=True, text=True).stdout.strip()
    with open(f"{REPO}/requirements.txt") as f:
        reqs = [l.strip() for l in f if l.strip() and not l.startswith("#")]
    keep = [r for r in reqs if r.split("==")[0].lower() not in {"torch", "jupyter", "ipykernel"}]
    run([sys.executable, "-m", "pip", "install", "-q", *keep])
INFO["cpu"] = cpu_model()
INFO["cpu_count"] = os.cpu_count()
save_info()
print("Commit:", INFO["commit"], "| CPU:", INFO["cpu"])
```

## pytest -q (dừng nếu lỗi)

```python
with step("pytest"):
    r = run([sys.executable, "-m", "pytest", "-q"], cwd=REPO)
    INFO["pytest"] = r.stdout.strip().splitlines()[-1]
    r2 = run([sys.executable, "-m", "pytest", "-q", "tests/test_ahp.py"], cwd=REPO)
    INFO["pytest_ahp"] = r2.stdout.strip().splitlines()[-1]
save_info()
```

## AHP trên số test đã có và ma trận đã duyệt

```python
with step("ahp"):
    run([sys.executable, "-I", "src/ahp.py", "reports/tables/summary_test.csv", "docs/ahp_matrix.json", f"{WORK}/ahp_result.json"], cwd=REPO)
    run([sys.executable, "-I", "scripts/ahp_plots.py", f"{WORK}/ahp_result.json", WORK], cwd=REPO)
res = json.load(open(f"{WORK}/ahp_result.json", encoding="utf-8"))
assert res["consistency"]["ok"], "CR >= 0,10"
INFO["recommended"], INFO["cr"] = res["recommended"], res["consistency"]["cr"]
save_info()
print(json.dumps({k: res[k] for k in ("weights", "consistency", "ranking", "recommended", "statements", "top_flips_under_20pct")}, ensure_ascii=False, indent=1))
print(os.listdir(WORK))
```
