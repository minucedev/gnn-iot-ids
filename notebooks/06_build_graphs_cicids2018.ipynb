# 06. Dựng đồ thị NF-CSE-CIC-IDS2018-v3 (bộ thứ 3, task 1.4, 1.6)

Bộ thứ 3 theo yêu cầu "≥ 3 bộ dữ liệu". W = S = 5 s, chia `blocked` 10 khối, gap = 1 cửa sổ (giống D2, D3). Bộ này **không phải IoT** và **không có lớp Khac** (chỉ 4 lớp). Chạy trong tiến trình con để `pip` không làm hỏng import của kernel.

```python
import glob, os, subprocess, sys, time

def run(cmd, **kw):
    r = subprocess.run(cmd, capture_output=True, text=True, **kw)
    print(r.stdout[-5000:], end="")
    print(r.stderr[-2500:], end="", file=sys.stderr)
    print(f"[mã thoát {r.returncode}]")
    if r.returncode != 0:
        raise RuntimeError(f"Lệnh lỗi: {' '.join(cmd)}")

REPO = "/tmp/gnn-iot-ids"
run(["rm", "-rf", REPO])
run(["git", "clone", "--depth", "1", "https://github.com/minucedev/gnn-iot-ids.git", REPO])
run(["git", "-C", REPO, "log", "-1", "--format=%H %s"])
with open(f"{REPO}/requirements.txt") as f:
    reqs = [l.strip() for l in f if l.strip() and not l.startswith("#")]
keep = [r for r in reqs if r.split("==")[0].lower() not in {"torch", "jupyter", "ipykernel"}]
run([sys.executable, "-m", "pip", "install", "-q", *keep])
```

```python
CSV = next(glob.iglob("/kaggle/input/**/NF-CICIDS2018-v3.csv", recursive=True))
print(CSV, os.path.getsize(CSV), "byte")
t0 = time.time()
run([sys.executable, "-I", "scripts/build_dataset_cli.py", "nf_cicids2018_v3", CSV, "5", "5", "blocked", "1",
     "/kaggle/working/processed/nf_cicids2018_v3", "nf_cicids2018_v3_w5s5_blocked"], cwd=REPO)
print(f"xong sau {time.time() - t0:.0f} s")
```
