# 03. Số node mỗi lớp theo cách chia tập, NF-ToN-IoT-v3 (số liệu cho D3, task 0.2)

Dựng bảng node/cạnh với W = S = 5 s bằng `src/build_graph.py`, rồi đếm số node mỗi lớp ở train/val/test cho `chronological` và `blocked` (10 khối). Chạy trong tiến trình con để `pip` không làm hỏng import của kernel.

```python
import glob, os, subprocess, sys

def run(cmd, **kw):
    r = subprocess.run(cmd, capture_output=True, text=True, **kw)
    print(r.stdout[-6000:], end="")
    print(r.stderr[-3000:], end="", file=sys.stderr)
    print(f"[mã thoát {r.returncode}]")
    if r.returncode != 0:
        raise RuntimeError(f"Lệnh lỗi: {' '.join(cmd)}")

REPO = "/kaggle/working/gnn-iot-ids"
run(["rm", "-rf", REPO])
run(["git", "clone", "--depth", "1", "https://github.com/minucedev/gnn-iot-ids.git", REPO])
run(["git", "-C", REPO, "log", "-1", "--format=%H %s"])
with open(f"{REPO}/requirements.txt") as f:
    reqs = [l.strip() for l in f if l.strip() and not l.startswith("#")]
keep = [r for r in reqs if r.split("==")[0].lower() not in {"torch", "jupyter", "ipykernel"}]
run([sys.executable, "-m", "pip", "install", "-q", *keep])
```

```python
CSV = next(f for f in glob.glob("/kaggle/input/**/NF-ToN-IoT-v3.csv", recursive=True))
print(CSV, os.path.getsize(CSV), "byte")
run([sys.executable, "-I", "scripts/split_stats.py", "nf_ton_iot_v3", CSV, "5", "5",
     "/kaggle/working/split_stats_nf_ton_iot_v3_w5.json"], cwd=REPO)
```
